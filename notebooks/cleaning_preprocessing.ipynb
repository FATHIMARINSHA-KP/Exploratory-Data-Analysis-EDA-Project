# DATA CLEANING & PREPROCESSING


```python
import pandas as pd 

import numpy as np

import matplotlib.pyplot as plt 

import seaborn as sns

df = pd.read_csv(r"C:\Users\A S S A U L T\Desktop\eda - project\data\raw\customer_churn_dataset.csv")

df.head()
```

### Fix Datatypes

```python
# Converted date_of_registration to datetime

df['date_of_registration']  = pd.to_datetime(df['date_of_registration'],dayfirst=True)

print(df['date_of_registration'])

df['estimated_salary'] = pd.to_numeric(df['estimated_salary'], errors='coerce')
```

### Standardize categorical variables

```python
# capitalize all string columns and trim white spaces 

for cols in df.select_dtypes(include=['object', 'string']).columns:
    
    df[cols] = df[cols].str.capitalize().str.strip()

df.head()
```

```python
# Standardize gender column

df['gender'] = df['gender'].str.upper().str.strip()

# Standardize telecom_partner column

df['telecom_partner'] = (df['telecom_partner'].str.strip().str.title())

df['telecom_partner'] = df['telecom_partner'].replace({'Bsnl': 'BSNL',})

df.head()
```

```python
# remove duplicates

df.drop_duplicates(inplace=True)

df.duplicated().sum()
```

### Handle missing values

```python
# identifying nulls

df.isnull().sum()
```

```python
# Handle missing values

# Fill missing values in age column using median

df['age'] = df['age'].fillna(df['age'].median())

# Fill missing values in state column using mode

df['state'] = df['state'].fillna(df['state'].mode()[0]) 

# Fill missing values in city column using mode

df['city'] = df['city'].fillna(df['city'].mode()[0]) 

# Fill missing values in estimated_salary column using median

df['estimated_salary'] = df['estimated_salary'].fillna(df['estimated_salary'].median())

#Fill missing values in sms_sent column using median

df['sms_sent'] = df['sms_sent'].fillna( df['sms_sent'].median())

#Fill missing values in data_used column using median

df['data_used'] = df['data_used'].fillna( df['data_used'].median())
```

```python
df.isnull().sum()
```

### Outlier Detection using IQR 

```python

# OUTLIER DETECTION FOR estimated_salary USING IQR

Q1 = df['estimated_salary'].quantile(0.25)
Q3 = df['estimated_salary'].quantile(0.75)

IQR = Q3 - Q1

LOWER = Q1 - 1.5 * IQR
UPPER = Q3 + 1.5 * IQR

outliers = df[(df['estimated_salary'] < LOWER) | (df['estimated_salary'] > UPPER)]

print(outliers['estimated_salary'])
```

```python
# OUTLIER DETECTION FOR sms_sent USING IQR

Q1 = df['sms_sent'].quantile(0.25)
Q3 = df['sms_sent'].quantile(0.75)

IQR = Q3 - Q1

LOWER = Q1 - 1.5 * IQR
UPPER = Q3 + 1.5 * IQR

outliers = df[(df['sms_sent'] < LOWER) | (df['sms_sent'] > UPPER)]

print(outliers['sms_sent'])
```

```python
# “Log transformation was applied to the estimated_salary column to reduce skewness and minimize the impact of extreme outlier values

import numpy as np

df['estimated_salary_log'] = np.log1p(df['estimated_salary'])
```

```python
# The estimated_salary column was highly right-skewed with extreme values. A log transformation (log1p) was applied to reduce skewness and normalize the distribution. After transformation, the distribution appeared more balanced
```
